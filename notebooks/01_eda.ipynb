```python
import pandas as pd

df = pd.read_csv("../data/raw/all_cities_historical.csv")
df["date"] = pd.to_datetime(df["date"], utc=True)
df = df.set_index("date")

df["hour"] = df.index.hour
df["month"] = df.index.month
df["day_of_week"] = df.index.dayofweek

df.shape
```

```python
df.isna().sum()  #Checking Missing Data or NULLS
```

```python
df[df["wind_speed_10m"].isna()]
```

```python
df.groupby("city")["us_aqi"].apply(lambda col: col.isna().sum())
```

```python
df["us_aqi"].describe()
```

```python
df["us_aqi"].hist(bins=100, figsize=(15, 5)) # split the full range of AQI values into 100 buckets
```

```python
df.groupby("city")["us_aqi"].mean().sort_values(ascending=False).plot(kind="bar", figsize=(10, 4))# City Wise AQI Mean
```

```python
Q1 = df["us_aqi"].quantile(0.25)
Q3 = df["us_aqi"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df["us_aqi"] < lower_bound) | (df["us_aqi"] > upper_bound)] #outside the normal range
print(f"Number of outlier rows: {len(outliers)}")
print(f"Outlier bounds: {lower_bound:.1f} to {upper_bound:.1f}")
```

```python
df.boxplot(column="us_aqi", by="city", figsize=(12, 5), rot=45)
```

```python
monthly = df.groupby(["city", "month"])["us_aqi"].mean().unstack(level=0)
monthly.plot(figsize=(14, 6), title="Monthly Average AQI by City")  #Season Wise Data
```

```python
numeric_cols = ["us_aqi", "pm10", "pm2_5", "carbon_monoxide", "nitrogen_dioxide",
                 "sulphur_dioxide", "ozone", "temperature_2m", "wind_speed_10m",
                 "wind_direction_10m", "relative_humidity_2m", "surface_pressure",
                 "precipitation"]

correlation_matrix = df[numeric_cols].corr()
correlation_matrix["us_aqi"].sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
plt.imshow(correlation_matrix, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar()
plt.xticks(range(len(numeric_cols)), numeric_cols, rotation=90)
plt.yticks(range(len(numeric_cols)), numeric_cols)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()                                               # whole grid
```

## EDA Findings
1.Any columns with meaningful missing data? Which, and roughly how much?
Answer: No columns have any missing values — .isna().sum() returned 0 across all 349,200 rows for every column.

2.Which city has the worst average AQI? Which has the best?
Answer: Faisalabad has the worst average AQI (156.7), narrowly ahead of Lahore (153.6) — though Lahore's boxplot shows more          dramatic outlier spikes. Quetta has the best average AQI (74.0), consistent with its tight, low boxplot.

3.What's the correlation between wind speed and AQI — and does that match what you'd physically expect?
Answer: AQI correlations: wind speed = −0.268 (higher wind → lower AQI, matches physical expectation — wind disperses pollution). Relative humidity = +0.188 (higher humidity → slightly higher AQI, consistent with humid air promoting particle formation and often coinciding with stagnant, low-dispersion conditions). Both moderate, not extreme — a healthy sign the model has real physical signal to learn from, not a shortcut.

4.Does seasonality look the same across all 10 cities, or different?
Answer: Seasonality follows the same overall shape in nearly every city — AQI peaks in December–January and dips in spring —         matching the winter smog pattern from Q5. The difference between cities is intensity, not timing:
        Faisalabad/Gujranwala swing dramatically (Jan avg 233 vs. Apr avg ~115), while Quetta stays comparatively flat year-round (peaks mildly in July, not winter).

5.Did you find outliers that look like real events versus likely sensor errors?
Answer: Outliers cluster heavily in winter (93% fall in Nov–Jan), matching Pakistan's documented smog season — a real seasonal pattern, not sensor error. The single highest spike (Lahore, 537.5 AQI, May 2025) coincides with elevated wind speed, more consistent with a dust storm event than trapped winter smog. No evidence of random/scattered sensor glitches.
